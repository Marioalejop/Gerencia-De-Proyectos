#Resumen Video "Dominio de Gobernanza"

El video explica que la gobernanza es la forma en que se establecen reglas, responsabilidades y procesos para dirigir un proyecto y asegurarse de que este cumpla con los objetivos de la organización. No se trata solamente de controlar lo que hace el equipo, sino de establecer una estructura que permita tomar buenas decisiones, gestionar riesgos y mantener el proyecto alineado con la estrategia del negocio.

1. Marco de Gobernanza
Define la gobernanza como el conjunto de reglas y procesos que dan dirección a través de políticas para cumplir los objetivos estratégicos. Se presenta como la "estructura de integridad" del proyecto: una infraestructura operativa (sistemas, procesos) que debe estar alineada con la estrategia del negocio y dar resiliencia técnica para soportar cambios.

2. Roles Clave
Presenta los roles típicos de un equipo (especialmente en marcos ágiles/Scrum):

Patrocinador (Sponsor): proporciona recursos.
Product Owner: gestiona el backlog.
Scrum Master: elimina obstáculos.
Desarrolladores: crean el producto.

3. Toma de Decisiones
Explica que las decisiones del proyecto dependen de elementos interconectados: gestión de riesgos, cronograma, cultura organizacional, requisitos técnicos, y el propio sistema de toma de decisiones — todos alineados y medidos.

4. Autogobierno Ágil
Contrasta dos modelos de liderazgo:

Liderazgo descendente: jerárquico, con un único punto de responsabilidad.
Autoridad distribuida: típica de equipos ágiles, con trabajo colaborativo y decisiones tomadas localmente. Los equipos ágiles con límites definidos usan reuniones diarias (daily stand-ups) para autoorganizarse.

5. Plan de Gobernanza (tutorial paso a paso)

Políticas: definir reglas y estándares de cumplimiento.
Límites: establecer umbrales de riesgo y control.
