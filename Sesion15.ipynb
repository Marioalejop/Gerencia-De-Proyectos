#Resumen Video "Sastrería y Cierre"

1. Sastrería (Tailoring)
Se define como la adaptación del enfoque y los procesos de un proyecto para ajustarlos al entorno y al tipo de trabajo. Se presenta el espectro de enfoques de desarrollo:

* Predictivo (planificación detallada por adelantado)
Híbrido (combinación de ambos)
Adaptativo (iterativo, tipo ágil)

* La idea central: no existe un único método correcto; el equipo "corta a la medida" los procesos según las necesidades del proyecto.

2. Retrospectiva
Un taller para analizar el trabajo realizado y descubrir oportunidades de mejora, con 4 pasos:

Reflexión – pensar sobre la iteración
Identificación – detectar éxitos y fallos
Documentación – registrar las lecciones aprendidas
Transferencia – actualizar los activos/procesos de la organización

3. Cierre del proyecto
Actividades clave para cerrar formalmente:

* Finalizar contratos
* Confirmar la aceptación (del cliente/patrocinador)
* Reasignar recursos
* Archivar registros
* Transferir a operaciones (el producto pasa a la fase operativa)
