#Resumen del Video "Dominio Financiero"

El dominio financiero en proyectos de software, es decir, cómo justificar un proyecto técnico en el lenguaje del negocio (según los procesos del PMBOK).

1. Valor del proyecto — "supervivencia corporativa": La balanza entre lo que cuesta el proyecto y el valor que entrega; sin esa justificación, el proyecto no sobrevive a la aprobación.

2. CapEx vs OpEx — "construyendo la base": Diferencia entre inversión de capital (servidores propios, on-premise) y gasto operativo recurrente (nube, suscripciones), y cómo esa elección cambia la estructura de costos.

3. Métricas de viabilidad — "el idioma del valor": Tres indicadores:

  * VAN: valor total actual de una serie de flujos de caja futuros.
  * TIR: tasa de interés a la que el VAN es igual a cero.
  * ROI y recuperación (payback): rentabilidad y tiempo exacto para recuperar el costo inicial.

4. Modelo de flujo de caja — "clímax del inversor":
 Flujo de cuatro pasos: proyectar ingresos y ahorros operativos → estimar costos (CapEx y OpEx) → calcular el flujo de caja neto → aplicar la tasa de descuento para obtener el VAN. Lo ilustra con cifras de ejemplo: 250 % de retorno y recuperación de la inversión en 14 meses (mes 1 → mes 6 → mes 14).

5. Entregables clave — "la resolución": El paquete de documentos financieros que debe producir el proyecto (caso de negocio, modelo de costos, etc.).
