#Resumen Video "Mente En el Trabajo"

El cerebro humano evolucionó para sobrevivir, no para el entorno laboral moderno, y eso explica los sesgos, el desgaste (burnout) y la importancia de crear equipos donde la gente se sienta segura.

1. Mente evolutiva y sesgos cognitivos

El cerebro prioriza la "ruta de supervivencia" (rápida, basada en atajos) sobre la "ruta completa" (lenta y analítica).
Un sesgo cognitivo es un efecto psicológico que desvía el procesamiento mental y produce juicios inexactos por atajos. Se asocia a Kahneman y Tversky (1972).
Estos atajos afectan decisiones de liderazgo y pueden contribuir al burnout.

2. Burnout

La pregunta que lo introduce: ¿qué pasa cuando el estrés crónico supera la capacidad de adaptación?
Se define como un síndrome por estrés crónico laboral no manejado con éxito. Es un fenómeno ocupacional, no una condición médica (referencia a la OMS).
Tiene tres dimensiones:
Agotamiento: falta total de energía.
Distancia: sentimiento de cinismo.
Ineficacia: menor eficacia profesional.

3. Seguridad psicológica

Se presenta un dato de encuesta sobre clima de equipo positivo: 43% sí y 57% no.
Sin apoyo, los trabajadores se desconectan y temen interactuar.

4. Las cuatro zonas (modelo de Amy Edmondson)

El modelo cruza dos ejes: nivel de apoyo y nivel de desafío.
Se ilustran con imágenes: la zona de confort (sillón), la de ansiedad (tornado) y la de apatía.
La zona ideal, con alto apoyo y alto desafío, es la de aprendizaje: estado de flujo, gente energizada que asume riesgos y colabora.
El contraste visual es entre un ambiente hostil (barrera con púas) y uno colaborativo (apretón de manos).

5. Liderazgo que funciona

Habilidades clave: diálogo abierto, patrocinio, humildad y conciencia cultural.
Cierre: la seguridad psicológica empieza cuando los altos líderes modelan las conductas (ciclo de aprendizaje, "trabajo en progreso", apertura a distintas opiniones).
