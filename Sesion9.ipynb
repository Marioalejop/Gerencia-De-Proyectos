3. Conceptos base: la Actividad
Define la actividad como la porción de trabajo distinta y programada que se realiza para producir los entregables del proyecto — la unidad mínima del cronograma.

4. Relaciones de precedencia (PDM – Método de Diagramación por Precedencia)
Explica los 4 tipos de dependencia entre actividades:

* FS (Final a Inicio): la sucesora no puede empezar hasta que termine la predecesora.
* SS (Inicio a Inicio): la sucesora no puede empezar hasta que empiece la predecesora.
* FF (Final a Final) y SF (Inicio a Final) también se mencionan como parte del set completo de relaciones.

5. Adelantos y Retrasos (Leads & Lags)

* Adelanto (Lead): tiempo en que una actividad sucesora puede adelantarse respecto a su predecesora.
* Retraso (Lag): tiempo de espera obligatorio entre una sucesora y su predecesora.
