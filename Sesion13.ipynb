#Resumen Video Valor Ganado (EVM)

Usa la metáfora de pilotar una nave espacial. Un proyecto complejo necesita un panel de control (dashboard) preciso para saber el combustible, la velocidad y la trayectoria. La sesión se organiza en cuatro partes.

1. El diagnóstico (dónde estamos hoy)

Todo gira en torno al 1.0, el punto de equilibrio. Por encima de 1.0 el proyecto va bien, y por debajo es una alerta roja de sobrecostos o retrasos.
CPI (índice de rendimiento de costos) = EV / AC. Es el medidor de combustible. Un CPI de 0.8 significa que por cada dólar gastado solo se obtienen 80 centavos de valor.
SPI (índice de rendimiento del cronograma) = EV / PV. Es el velocímetro. Si es menor a 1.0, el proyecto va más lento de lo planificado.

2. La bola de cristal (pronósticos)

EAC (estimación al finalizar): el costo final proyectado si se mantiene el rendimiento actual.
ETC (estimación para completar): el dinero adicional necesario desde hoy para terminar.
VAC (variación al finalizar) = BAC − EAC. Indica si el proyecto terminará con superávit o déficit. Un VAC negativo señala que las reservas de contingencia están en riesgo.

3. El esfuerzo requerido (TCPI)

El TCPI es el trabajo que falta dividido entre los fondos que quedan. Muestra la eficiencia que el equipo debe sostener desde la fecha de corte para cumplir el presupuesto original.
Si ese presupuesto ya es inviable, se recalcula el TCPI usando el EAC como base. Así se negocia un presupuesto nuevo y realista.

4. La síntesis (dashboard ejecutivo)

Se construye en tres pasos:
Graficar las curvas S de valor planificado, costo real y valor ganado.
Integrar el diagnóstico, con las brechas de costo y tiempo codificadas por colores frente al 1.0.
Trazar la proyección con el EAC y el TCPI.
Debe incluir las curvas EVM actualizadas, CPI y SPI, EAC y VAC, y el TCPI como meta operativa.

Los proyectos rara vez se descontrolan de golpe. Suelen deteriorarse por pequeñas ineficiencias acumuladas en CPI y SPI que nadie vio. Dominar el valor ganado es asegurar resultados, no solo calcular métricas.

#Resumen Video Gestión de Riesgos

Se basa en la sección 2.7 del PMBOK (dominio de desempeño del riesgo) y usa un caso de software. La pregunta central es cuánto pierde el proyecto si un proveedor externo de API colapsa. La sesión tiene seis secciones.

1. Introducción al riesgo

Un riesgo es un evento incierto que afecta los objetivos del proyecto.
En software no es solo un error de código. Es cualquier cosa que impacte los tiempos de entrega y, por tanto, los presupuestos (CAPEX y OPEX).

2. Análisis cualitativo

Es un mapa rápido tipo semáforo, basado en el juicio de expertos.
Su herramienta clave es la matriz de probabilidad-impacto, que prioriza los riesgos de muy bajo a muy alto y ayuda a enfocar recursos en las amenazas críticas.

3. Análisis cuantitativo

Es el "GPS financiero", basado en datos y simulaciones.
El árbol de decisiones compara caminos por su valor esperado. Por ejemplo, invertir 50M en una plataforma nueva o actualizar y licenciar la actual.
El diagrama de tornado muestra qué variables tienen mayor impacto, con las barras más largas arriba.
La simulación de Monte Carlo corre miles de escenarios a la vez. Calcula la probabilidad de cumplir los plazos y proteger la línea base.

4. Tutorial del Valor Monetario Esperado (VME)

Escenario: retraso del proveedor de la API.
Probabilidad de 60% e impacto de $50.000 (desarrolladores de brazos cruzados).
VME = 0.60 × $50.000 = $30.000. Es la reserva de contingencia que debe bloquearse en el presupuesto.

5. Estrategias de respuesta

Mitigar: construir un componente de respaldo para no depender al 100% de la API.
Aceptar: documentar la amenaza, guardar la reserva de $30.000 y monitorear.
Escalar: pasar el riesgo a la alta dirección cuando su impacto supera la autoridad del proyecto.

6. Registro de riesgos (risk register)

Es el entregable que consolida todo: el evento, la probabilidad (60%), el impacto ($50.000), la reserva ($30.000), la estrategia de mitigación y el responsable (el líder técnico). Da rendición de cuentas y control.
