#Resumen del video "Gerencia de Proyectos y Valor"


El video habla de gestión de proyectos de software desde la óptica de la nueva Guía PMBOK, la octava edición, y lo hace en cinco bloques.

Se compara el antes y el después: antes la gestión de proyectos era un enredo de procesos rígidos, y ahora la idea es un enfoque mucho más simple, donde todo gira en torno a generar valor real para el negocio.

Después entra en cómo encajan los proyectos dentro de una empresa. No es solo "un proyecto de software" aislado, sino que hace parte de un ecosistema más grande: programas, portafolios, toda la organización. Ahí suelta una idea que resume bastante bien el punto: la gestión de portafolio es lo que conecta la estrategia de la empresa con lo que realmente se ejecuta en el día a día.

La tercera parte se centra en el valor. Dice algo que vale la pena quedarse con eso: el verdadero indicador de éxito de un proyecto no es que se entregue a tiempo o dentro del presupuesto, sino cuánto valor genera por cada peso invertido.

Luego pasa a un tutorial práctico de viabilidad, en tres pasos: primero identificar la necesidad u oportunidad de negocio, después definir objetivos que se puedan medir, y por último dejar claro cómo se va a saber si el proyecto fue un éxito.

Cierra con los documentos formales: el Caso de Negocio (que justifica por qué vale la pena hacer el proyecto, viendo necesidades, alineación estratégica y beneficios financieros) y el Acta Constitutiva (que ya lo vuelve oficial, con objetivos medibles, alcance, recursos y criterios de cierre).

Y termina con una pregunta que deja picando: si la tecnología (y ahí mete la IA) está pensada solo para escribir código, o para transformar de verdad el negocio.
